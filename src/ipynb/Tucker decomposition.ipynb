```python
import torch.nn.functional as F
from sentence_transformers import SentenceTransformer
from transformers import AutoImageProcessor, AutoModel
import torch
from sentence_transformers import util

device = "cuda" if torch.cuda.is_available() else "cpu"

# load models once
clip32_model = SentenceTransformer('clip-ViT-B-32', device=device)
dino_processor = AutoImageProcessor.from_pretrained('facebook/dinov2-giant')
dino_model = AutoModel.from_pretrained(
    'facebook/dinov2-giant',
    torch_dtype=torch.bfloat16 if device == "cuda" else torch.float32
).to(device)

# SentenceTransformer device
print("CLIP device:", clip32_model.device)

# DINO model device
print("DINO device:", next(dino_model.parameters()).device)

# Optional: verify CUDA
if torch.cuda.is_available():
  print("Current CUDA device:", torch.cuda.current_device())
  print("GPU name:", torch.cuda.get_device_name(0))
```

```python
def clip_IS_emb(imgs):
  # clip vitb/32
  image_embs = clip32_model.encode(imgs, convert_to_tensor=True)

  return image_embs

def clip_IS_score(real_embs, gen_embs):
  clip32_IS = util.cos_sim(real_embs, gen_embs).mean().item()

  return clip32_IS

def dino_IS_emb(imgs):
  # pre-process
  inputs = dino_processor(images=imgs, return_tensors="pt").to(dino_model.device)

  # run
  with torch.no_grad():
      outputs = dino_model(**inputs)

  # extract embeddings (right before the classification layer)
  image_embs = outputs.last_hidden_state[:, 0]

  return image_embs

def dino_IS_score(real_emb, gen_emb):
  dino_IS = util.cos_sim(real_emb, gen_emb).mean().item()

  return dino_IS

def TS_emb(image_or_prompt):
  # clip vitb/32
  emb = clip32_model.encode(image_or_prompt, convert_to_tensor=True)

  return emb

def TS_score(image_emb, text_emb):
  text_similarity = util.cos_sim(image_emb, text_emb).mean().item()

  return text_similarity
```

###### Calculations

```python
import numpy as np
import pandas as pd
from datetime import datetime

clip32_IS = {}
dino_IS = {}
clip_real_embs = {}
dino_real_embs = {}

# embed for real images
print("Real images embed")
real_classes = sorted(real_images.keys())
for real_cls in real_classes:
  if real_cls not in clip_real_embs:
    clip_real_embs[real_cls] = clip_IS_emb(real_images[real_cls])
  if real_cls not in dino_real_embs:
    dino_real_embs[real_cls] = dino_IS_emb(real_images[real_cls])

print("Gen images")
# embed for gen images and final calc
for model in gen_images.keys():
  clip32_IS[model] = {}
  dino_IS[model] = {}
  for checkpoint in gen_images[model].keys():
    clip32_IS[model][checkpoint] = {}
    dino_IS[model][checkpoint] = {}

    for infer_step in gen_images[model][checkpoint].keys():
      gen_classes = sorted(gen_images[model][checkpoint][infer_step].keys())

      # clip32
      clip32_matrix = pd.DataFrame(index=real_classes, columns=gen_classes)

      for real_cls in real_classes:
        for gen_cls in gen_classes:
          gen_embs = clip_IS_emb(gen_images[model][checkpoint][infer_step][gen_cls])

          clip32_matrix.loc[real_cls, gen_cls] = clip_IS_score(clip_real_embs[real_cls], gen_embs)

        clip32_IS[model][checkpoint][infer_step] = clip32_matrix

      clip32_path = os.path.join(scores_dir, f"clipIS_{model}_{checkpoint}_{infer_step}.csv")
      clip32_IS[model][checkpoint][infer_step].to_csv(clip32_path)
      print(f"Clip32 IS for model {model}, {checkpoint}, {infer_step} done. {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")

      # dino
      dino_matrix = pd.DataFrame(index=real_classes, columns=gen_classes)

      for real_cls in real_classes:
        for gen_cls in gen_classes:
          gen_embs = dino_IS_emb(gen_images[model][checkpoint][infer_step][gen_cls])

          dino_matrix.loc[real_cls, gen_cls] = dino_IS_score(dino_real_embs[real_cls], gen_embs)

      dino_IS[model][checkpoint][infer_step] = dino_matrix

      dino_path = os.path.join(scores_dir, f"dinoIS_{model}_{checkpoint}_{infer_step}.csv")
      dino_IS[model][checkpoint][infer_step].to_csv(dino_path)
      print(f"Dino IS for model {model}, {checkpoint}, {infer_step} done. {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")
```

```python
from collections import defaultdict
from datetime import datetime
score_dict = {}
for idx, (sample_path, model_name) in enumerate(zip(sample_paths, model_paths)):
  if model_name not in score_dict:
    score_dict[model_name] = {}
  for checkpoint in checkpoint_list:
    score_dict[model_name][f"checkpoint-{checkpoint}"] = {}
    for infer_step in gen_images[model_name][f"checkpoint-{checkpoint}"].keys():
      score_dict[model_name][f"checkpoint-{checkpoint}"][infer_step] = {}
      for class_name in gen_images[model_name][f"checkpoint-{checkpoint}"][infer_step].keys():
        class_scores = []
        for prompt, images in gen_images[model_name][f"checkpoint-{checkpoint}"][infer_step][class_name].items():
          TS_prompt_image = TS_score(TS_emb(images), TS_emb(prompt))
          class_scores.append(TS_prompt_image)

      print(f"TS done for {model_name}, {checkpoint}, {infer_step}. {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")
      score_dict[model_name][f"checkpoint-{checkpoint}"][infer_step][class_name] = sum(class_scores) / len(class_scores)
```

```python
# calculate checkpoint-wise average
import json, os
avg_score = {}
for model_name in score_dict.keys():
  avg_score[model_name] = {}
  for checkpoint in score_dict[model_name].keys():
    avg_score[model_name][checkpoint] = {}
    for infer_step in score_dict[model_name][checkpoint].keys():
      if len(score_dict[model_name][checkpoint][infer_step]) == 0:
        avg_score[model_name][checkpoint][infer_step] = "N/A"
      else:
        avg_score[model_name][checkpoint][infer_step] = \
        sum(score_dict[model_name][checkpoint][infer_step].values()) / \
        len(score_dict[model_name][checkpoint][infer_step])

print(json.dumps(avg_score, indent=2))
```

```python
import torch.nn.functional as F
from sentence_transformers import SentenceTransformer
from transformers import AutoImageProcessor, AutoModel
import torch
from sentence_transformers import util

device = "cuda" if torch.cuda.is_available() else "cpu"

# load models once
clip32_model = SentenceTransformer('clip-ViT-B-32', device=device)
dino_processor = AutoImageProcessor.from_pretrained('facebook/dinov2-giant')
dino_model = AutoModel.from_pretrained(
    'facebook/dinov2-giant',
    torch_dtype=torch.bfloat16 if device == "cuda" else torch.float32
).to(device)

# SentenceTransformer device
print("CLIP device:", clip32_model.device)

# DINO model device
print("DINO device:", next(dino_model.parameters()).device)

# Optional: verify CUDA
if torch.cuda.is_available():
  print("Current CUDA device:", torch.cuda.current_device())
  print("GPU name:", torch.cuda.get_device_name(0))
```

```python
def clip_IS_emb(imgs):
  # clip vitb/32
  image_embs = clip32_model.encode(imgs, convert_to_tensor=True)

  return image_embs

def clip_IS_score(real_embs, gen_embs):
  clip32_IS = util.cos_sim(real_embs, gen_embs).mean().item()

  return clip32_IS

def dino_IS_emb(imgs):
  # pre-process
  inputs = dino_processor(images=imgs, return_tensors="pt").to(dino_model.device)

  # run
  with torch.no_grad():
      outputs = dino_model(**inputs)

  # extract embeddings (right before the classification layer)
  image_embs = outputs.last_hidden_state[:, 0]

  return image_embs

def dino_IS_score(real_emb, gen_emb):
  dino_IS = util.cos_sim(real_emb, gen_emb).mean().item()

  return dino_IS

def TS_emb(image_or_prompt):
  # clip vitb/32
  emb = clip32_model.encode(image_or_prompt, convert_to_tensor=True)

  return emb

def TS_score(image_emb, text_emb):
  text_similarity = util.cos_sim(image_emb, text_emb).mean().item()

  return text_similarity
```

```python
import torch.nn.functional as F
from sentence_transformers import SentenceTransformer
from transformers import AutoImageProcessor, AutoModel
import torch
from sentence_transformers import util

device = "cuda" if torch.cuda.is_available() else "cpu"

# load models once
clip32_model = SentenceTransformer('clip-ViT-B-32', device=device)
dino_processor = AutoImageProcessor.from_pretrained('facebook/dinov2-giant')
dino_model = AutoModel.from_pretrained(
    'facebook/dinov2-giant',
    torch_dtype=torch.bfloat16 if device == "cuda" else torch.float32
).to(device)

# SentenceTransformer device
print("CLIP device:", clip32_model.device)

# DINO model device
print("DINO device:", next(dino_model.parameters()).device)

# Optional: verify CUDA
if torch.cuda.is_available():
  print("Current CUDA device:", torch.cuda.current_device())
  print("GPU name:", torch.cuda.get_device_name(0))
```

```python
def clip_IS_emb(imgs):
  # clip vitb/32
  image_embs = clip32_model.encode(imgs, convert_to_tensor=True)

  return image_embs

def clip_IS_score(real_embs, gen_embs):
  clip32_IS = util.cos_sim(real_embs, gen_embs).mean().item()

  return clip32_IS

def dino_IS_emb(imgs):
  # pre-process
  inputs = dino_processor(images=imgs, return_tensors="pt").to(dino_model.device)

  # run
  with torch.no_grad():
      outputs = dino_model(**inputs)

  # extract embeddings (right before the classification layer)
  image_embs = outputs.last_hidden_state[:, 0]

  return image_embs

def dino_IS_score(real_emb, gen_emb):
  dino_IS = util.cos_sim(real_emb, gen_emb).mean().item()

  return dino_IS

def TS_emb(image_or_prompt):
  # clip vitb/32
  emb = clip32_model.encode(image_or_prompt, convert_to_tensor=True)

  return emb

def TS_score(image_emb, text_emb):
  text_similarity = util.cos_sim(image_emb, text_emb).mean().item()

  return text_similarity
```

```python
import numpy as np
import pandas as pd
from datetime import datetime

clip32_IS = {}
dino_IS = {}
clip_real_embs = {}
dino_real_embs = {}

# embed for real images
print("Real images embed")
real_classes = sorted(real_images.keys())
for real_cls in real_classes:
  if real_cls not in clip_real_embs:
    clip_real_embs[real_cls] = clip_IS_emb(real_images[real_cls])
  if real_cls not in dino_real_embs:
    dino_real_embs[real_cls] = dino_IS_emb(real_images[real_cls])

print("Gen images")
# embed for gen images and final calc
for model in gen_images.keys():
  clip32_IS[model] = {}
  dino_IS[model] = {}
  for checkpoint in gen_images[model].keys():
    clip32_IS[model][checkpoint] = {}
    dino_IS[model][checkpoint] = {}

    for infer_step in gen_images[model][checkpoint].keys():
      gen_classes = sorted(gen_images[model][checkpoint][infer_step].keys())

      # clip32
      clip32_matrix = pd.DataFrame(index=real_classes, columns=gen_classes)

      for real_cls in real_classes:
        for gen_cls in gen_classes:
          gen_embs = clip_IS_emb(gen_images[model][checkpoint][infer_step][gen_cls])

          clip32_matrix.loc[real_cls, gen_cls] = clip_IS_score(clip_real_embs[real_cls], gen_embs)

        clip32_IS[model][checkpoint][infer_step] = clip32_matrix

      clip32_path = os.path.join(scores_dir, f"clipIS_{model}_{checkpoint}_{infer_step}.csv")
      clip32_IS[model][checkpoint][infer_step].to_csv(clip32_path)
      print(f"Clip32 IS for model {model}, {checkpoint}, {infer_step} done. {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")

      # dino
      dino_matrix = pd.DataFrame(index=real_classes, columns=gen_classes)

      for real_cls in real_classes:
        for gen_cls in gen_classes:
          gen_embs = dino_IS_emb(gen_images[model][checkpoint][infer_step][gen_cls])

          dino_matrix.loc[real_cls, gen_cls] = dino_IS_score(dino_real_embs[real_cls], gen_embs)

      dino_IS[model][checkpoint][infer_step] = dino_matrix

      dino_path = os.path.join(scores_dir, f"dinoIS_{model}_{checkpoint}_{infer_step}.csv")
      dino_IS[model][checkpoint][infer_step].to_csv(dino_path)
      print(f"Dino IS for model {model}, {checkpoint}, {infer_step} done. {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")
```

```python
import torch.nn.functional as F
from sentence_transformers import SentenceTransformer
from transformers import AutoImageProcessor, AutoModel
import torch
from sentence_transformers import util

device = "cuda" if torch.cuda.is_available() else "cpu"

# load models once
clip32_model = SentenceTransformer('clip-ViT-B-32', device=device)
dino_processor = AutoImageProcessor.from_pretrained('facebook/dinov2-giant')
dino_model = AutoModel.from_pretrained(
    'facebook/dinov2-giant',
    torch_dtype=torch.bfloat16 if device == "cuda" else torch.float32
).to(device)

# SentenceTransformer device
print("CLIP device:", clip32_model.device)

# DINO model device
print("DINO device:", next(dino_model.parameters()).device)

# Optional: verify CUDA
if torch.cuda.is_available():
  print("Current CUDA device:", torch.cuda.current_device())
  print("GPU name:", torch.cuda.get_device_name(0))
```

```python
def clip_IS_emb(imgs):
  # clip vitb/32
  image_embs = clip32_model.encode(imgs, convert_to_tensor=True)

  return image_embs

def clip_IS_score(real_embs, gen_embs):
  clip32_IS = util.cos_sim(real_embs, gen_embs).mean().item()

  return clip32_IS

def dino_IS_emb(imgs):
  # pre-process
  inputs = dino_processor(images=imgs, return_tensors="pt").to(dino_model.device)

  # run
  with torch.no_grad():
      outputs = dino_model(**inputs)

  # extract embeddings (right before the classification layer)
  image_embs = outputs.last_hidden_state[:, 0]

  return image_embs

def dino_IS_score(real_emb, gen_emb):
  dino_IS = util.cos_sim(real_emb, gen_emb).mean().item()

  return dino_IS

def TS_emb(image_or_prompt):
  # clip vitb/32
  emb = clip32_model.encode(image_or_prompt, convert_to_tensor=True)

  return emb

def TS_score(image_emb, text_emb):
  text_similarity = util.cos_sim(image_emb, text_emb).mean().item()

  return text_similarity
```

```python
import torch.nn.functional as F
from sentence_transformers import SentenceTransformer
from transformers import AutoImageProcessor, AutoModel
import torch
from sentence_transformers import util

device = "cuda" if torch.cuda.is_available() else "cpu"

# load models once
clip32_model = SentenceTransformer('clip-ViT-B-32', device=device)
dino_processor = AutoImageProcessor.from_pretrained('facebook/dinov2-giant')
dino_model = AutoModel.from_pretrained(
    'facebook/dinov2-giant',
    torch_dtype=torch.bfloat16 if device == "cuda" else torch.float32
).to(device)

# SentenceTransformer device
print("CLIP device:", clip32_model.device)

# DINO model device
print("DINO device:", next(dino_model.parameters()).device)

# Optional: verify CUDA
if torch.cuda.is_available():
  print("Current CUDA device:", torch.cuda.current_device())
  print("GPU name:", torch.cuda.get_device_name(0))
```

```python
def clip_IS_emb(imgs):
  # clip vitb/32
  image_embs = clip32_model.encode(imgs, convert_to_tensor=True)

  return image_embs

def clip_IS_score(real_embs, gen_embs):
  clip32_IS = util.cos_sim(real_embs, gen_embs).mean().item()

  return clip32_IS

def dino_IS_emb(imgs):
  # pre-process
  inputs = dino_processor(images=imgs, return_tensors="pt").to(dino_model.device)

  # run
  with torch.no_grad():
      outputs = dino_model(**inputs)

  # extract embeddings (right before the classification layer)
  image_embs = outputs.last_hidden_state[:, 0]

  return image_embs

def dino_IS_score(real_emb, gen_emb):
  dino_IS = util.cos_sim(real_emb, gen_emb).mean().item()

  return dino_IS

def TS_emb(image_or_prompt):
  # clip vitb/32
  emb = clip32_model.encode(image_or_prompt, convert_to_tensor=True)

  return emb

def TS_score(image_emb, text_emb):
  text_similarity = util.cos_sim(image_emb, text_emb).mean().item()

  return text_similarity
```

```python
import numpy as np
import pandas as pd
from datetime import datetime

clip32_IS = {}
dino_IS = {}
clip_real_embs = {}
dino_real_embs = {}

# embed for real images
print("Real images embed")
real_classes = sorted(real_images.keys())
for real_cls in real_classes:
  if real_cls not in clip_real_embs:
    clip_real_embs[real_cls] = clip_IS_emb(real_images[real_cls])
  if real_cls not in dino_real_embs:
    dino_real_embs[real_cls] = dino_IS_emb(real_images[real_cls])

print("Gen images")
# embed for gen images and final calc
for model in gen_images.keys():
  clip32_IS[model] = {}
  dino_IS[model] = {}
  for checkpoint in gen_images[model].keys():
    clip32_IS[model][checkpoint] = {}
    dino_IS[model][checkpoint] = {}

    for infer_step in gen_images[model][checkpoint].keys():
      gen_classes = sorted(gen_images[model][checkpoint][infer_step].keys())

      # clip32
      clip32_matrix = pd.DataFrame(index=real_classes, columns=gen_classes)

      for real_cls in real_classes:
        for gen_cls in gen_classes:
          gen_embs = clip_IS_emb(gen_images[model][checkpoint][infer_step][gen_cls])

          clip32_matrix.loc[real_cls, gen_cls] = clip_IS_score(clip_real_embs[real_cls], gen_embs)

        clip32_IS[model][checkpoint][infer_step] = clip32_matrix

      clip32_path = os.path.join(scores_dir, f"clipIS_{model}_{checkpoint}_{infer_step}.csv")
      clip32_IS[model][checkpoint][infer_step].to_csv(clip32_path)
      print(f"Clip32 IS for model {model}, {checkpoint}, {infer_step} done. {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")

      # dino
      dino_matrix = pd.DataFrame(index=real_classes, columns=gen_classes)

      for real_cls in real_classes:
        for gen_cls in gen_classes:
          gen_embs = dino_IS_emb(gen_images[model][checkpoint][infer_step][gen_cls])

          dino_matrix.loc[real_cls, gen_cls] = dino_IS_score(dino_real_embs[real_cls], gen_embs)

      dino_IS[model][checkpoint][infer_step] = dino_matrix

      dino_path = os.path.join(scores_dir, f"dinoIS_{model}_{checkpoint}_{infer_step}.csv")
      dino_IS[model][checkpoint][infer_step].to_csv(dino_path)
      print(f"Dino IS for model {model}, {checkpoint}, {infer_step} done. {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")
```

###### Calculate

```python
from collections import defaultdict
from datetime import datetime
score_dict = {}
for idx, (sample_path, model_name) in enumerate(zip(sample_paths, model_paths)):
  if model_name not in score_dict:
    score_dict[model_name] = {}
  for checkpoint in checkpoint_list:
    score_dict[model_name][f"checkpoint-{checkpoint}"] = {}
    for infer_step in gen_images[model_name][f"checkpoint-{checkpoint}"].keys():
      score_dict[model_name][f"checkpoint-{checkpoint}"][infer_step] = {}
      for class_name in gen_images[model_name][f"checkpoint-{checkpoint}"][infer_step].keys():
        class_scores = []
        for prompt, images in gen_images[model_name][f"checkpoint-{checkpoint}"][infer_step][class_name].items():
          TS_prompt_image = TS_score(TS_emb(images), TS_emb(prompt))
          class_scores.append(TS_prompt_image)

      print(f"TS done for {model_name}, {checkpoint}, {infer_step}. {datetime.now().strftime("%Y-%m-%d %H:%M:%S")}")
      score_dict[model_name][f"checkpoint-{checkpoint}"][infer_step][class_name] = sum(class_scores) / len(class_scores)
```

```python
# calculate checkpoint-wise average
import json, os
avg_score = {}
for model_name in score_dict.keys():
  avg_score[model_name] = {}
  for checkpoint in score_dict[model_name].keys():
    avg_score[model_name][checkpoint] = {}
    for infer_step in score_dict[model_name][checkpoint].keys():
      if len(score_dict[model_name][checkpoint][infer_step]) == 0:
        avg_score[model_name][checkpoint][infer_step] = "N/A"
      else:
        avg_score[model_name][checkpoint][infer_step] = \
        sum(score_dict[model_name][checkpoint][infer_step].values()) / \
        len(score_dict[model_name][checkpoint][infer_step])

print(json.dumps(avg_score, indent=2))
```

```python
output_file_path = os.path.join(scores_dir, f"TS_scores.json")

if not os.path.exists(output_file_path):
  with open(output_file_path, 'w') as f:
    json.dump(avg_score, f, indent=2)
else:
  previous_data = {}
  with open(output_file_path, 'r+') as f:
    previous_data = json.load(f)

    for model_name in avg_score.keys():
      if model_name not in previous_data:
        previous_data[model_name] = {}
      for checkpoint in avg_score[model_name].keys():
        if checkpoint not in previous_data[model_name]:
          previous_data[model_name][checkpoint] = {}
        for infer_step in avg_score[model_name][checkpoint].keys():
            previous_data[model_name][checkpoint][infer_step] = avg_score[model_name][checkpoint][infer_step]

    f.seek(0)
    json.dump(previous_data, f, indent=2)
    f.truncate()
```
